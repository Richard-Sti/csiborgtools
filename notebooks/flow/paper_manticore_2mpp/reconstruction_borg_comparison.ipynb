```python
import matplotlib.lines as mlines
import matplotlib.patches as mpatches
import matplotlib.transforms as mtransforms


def make_colours(y, n, k):
    sorted_indices = np.argsort(y[n, k])[::-1]  # Sort in descending order
    # Initialize colors to gray
    colors = ['olivedrab'] * len(y[n, k])
    markers = ['X'] * len(y[n, k])

    # Assign specific colors to top 3 values
    if len(sorted_indices) >= 1:
        colors[sorted_indices[0]] = '#FFD700'  # Highest value gets gold
        markers[sorted_indices[0]] = 'o'
    if len(sorted_indices) >= 2:
        colors[sorted_indices[1]] = '#C0C0C0'  # Second highest gets silver
        markers[sorted_indices[1]] = '8'
    if len(sorted_indices) >= 3:
        colors[sorted_indices[2]] = "#CD7F32"  # Third highest gets bronze
        markers[sorted_indices[2]] = 'p'
    return colors, markers


with plt.style.context('science'):
    plt.rcParams.update({'font.size': 9})
    figwidth = 8.3
    fig, axs = plt.subplots(2, 3, figsize=(figwidth, 0.55 * figwidth))
    fig.subplots_adjust(hspace=0, wspace=0)

    x = np.arange(len(sims))
    y = y_lnZ

    for n in range(len(catalogues)):
        i, j = n // 3, n % 3
        ax = axs[i, j]
        # twin = ax.twinx()
        # twin.tick_params(axis='y', which='minor', length=0)

        # if j == 2:
            # twin.set_ylabel(r"$\Delta \log_{10} \mathcal{Z}_{\rm smoothed}$")

        ax.text(0.025, 1.075, catalogue_to_pretty(catalogues[n]),
                transform=ax.transAxes,
                verticalalignment='center', horizontalalignment='left',
                bbox=dict(facecolor='white', alpha=0.85, edgecolor='none', pad=3),
                zorder=5)

        bar_width = 0.35
        colors, markers = make_colours(y, n, 0)
        for i in range(len(x)):
            ax.bar(x[i] - bar_width / 2, y[n, 0][i], width=0.35 * 1.25, color=colors[i], edgecolor="k", zorder=-1)

        # colors, markers = make_colours(y, n, 1)
        # for i in range(len(x)):
        #     twin.bar(x[i] + bar_width / 2, y[n, 1][i], width=0.35 * 0.75,
        #              color=colors[i], edgecolor="k", hatch='x', zorder=-1, alpha=0.25)


    for i in range(3):
        axs[1, i].set_xticks(
            np.arange(len(sims)),
            [simname_to_pretty(sim) for sim in sims], rotation=45,
            ha="right")

         # Apply a slight translation to move the labels rightward
        offset = mtransforms.ScaledTranslation(5 / 72, 0, fig.dpi_scale_trans)
        for label in axs[1, i].get_xticklabels():
            label.set_transform(label.get_transform() + offset)
        axs[0, i].set_xticks([], [])

    for i in range(2):
        for j in range(3):
            axs[i, j].set_xlim(-0.75, len(sims) - 0.25)

            axs[i, j].tick_params(axis='x', which='major', top=False)
            axs[i, j].tick_params(axis='x', which='minor', top=False, length=0)
            axs[i, j].tick_params(axis='y', which='minor', length=0)

        axs[i, 0].set_ylabel(r"$\Delta \log_{10} \mathcal{Z}$")

    gold_patch = mpatches.Patch(color='#FFD700', label=r"1\textsuperscript{st} place")
    silver_patch = mpatches.Patch(color='#C0C0C0', label=r"2\textsuperscript{nd} place")
    bronze_patch = mpatches.Patch(color='#CD7F32', label=r"3\textsuperscript{rd} place")
    green_patch = mpatches.Patch(color='olivedrab', label=r"Remaining")

    # Add the legend to the figure
    fig.legend(handles=[gold_patch, silver_patch, bronze_patch, green_patch], 
               loc='upper center', bbox_to_anchor=(0.4, 1.05), ncol=4,
               frameon=False)

    # # Add a separate information box for hatched and solid bars in the top right corner
    # hatched_text = "Solid bars: Highest-resolution fields\nHatched bars: Smoothed fields"
    # fig.text(0.725, 1.04, hatched_text, ha='left', va='top',
    #          bbox=dict(facecolor='white', edgecolor='gray', boxstyle='square,pad=0.3', alpha=0.85),
    #          transform=fig.transFigure)

    fig.tight_layout()
    fname = "../../../plots/BORG_logZ_comparison.pdf"
    # fig.savefig(fname, dpi=500, bbox_inches='tight')
    fig.show()
```

```python
sims = ["csiborg1", "csiborg2_main", "manticore_2MPP_MULTIBIN_N256_DES_V2"]
cats = ["2MTF", "SFI_gals", "CF4_TFR_i", "CF4_TFR_w1"]

X = {}

for sim in sims:
    for cat in cats:
        fname = paths.flow_validation(
            fdir, sim, cat, inference_method="bayes",
            sample_alpha=True, zcmb_max=0.05)

        if not exists(fname):
            raise FileNotFoundError(fname)

        with File(fname, 'r') as f:
            X[f"{sim}_{cat}"] = np.linalg.norm(f[f"samples/Vext"][...], axis=1)
```
